## Calcula
1) Altura promedio de los estudiantes varones:
2) Altura promedio de las estudiantes:
3) Peso promedio de los estudiantes varones:
4) Peso promedio de las estudiantes:
5) La mayor estatura entre los estudiantes varones:
6) La menor estatura entre las estudiantes:
7) El peso más ligero entre los estudiantes varones de estatura superior a 1.75 m:
8) El mayor peso entre las estudiantes de estatura inferior a 1.60 m:
9) La nota promedio de los alumnos (hombres y mujeres) sin ausencias ('N'):
10) La nota promedio de los alumnos (hombres y mujeres) con absentismo ('Y'):
11) Altura promedio de los alumnos (hombres y mujeres) con grupo sanguíneo 'A' o 'AB':
12) Altura promedio de los alumnos varones con grupo sanguíneo 'A' o 'AB':
13) Edad promedio de los alumnos (hombres y mujeres) con absentismo ('Y') cuya nota es mayor o igual a 3:

```python
import pandas as pd
import os

# Cargar DataFrame
df = pd.read_csv('../data/data_studentlist.csv', header='infer')

# 1. Altura promedio de los estudiantes varones
alt_prom_varones = df[df['Gender'] == 'M']['Height'].mean()
print(f"1. Altura promedio de los estudiantes varones: {alt_prom_varones:.2f}")

# 2. Altura promedio de las estudiantes
alt_prom_mujeres = df[df['Gender'] == 'F']['Height'].mean()
print(f"2. Altura promedio de las estudiantes: {alt_prom_mujeres:.2f}")

# 3. Peso promedio de los varones
peso_prom_varones = df[df['Gender'] == 'M']['Weight'].mean()
print(f"3. Peso promedio de los estudiantes varones: {peso_prom_varones:.2f}")

# 4. Peso promedio de las estudiantes
peso_prom_mujeres = df[df['Gender'] == 'F']['Weight'].mean()
print(f"4. Peso promedio de las estudiantes: {peso_prom_mujeres:.2f}")

# 5. La mayor estatura de los estudiantes varones
max_alt_varones = df[df['Gender'] == 'M']['Height'].max()
print(f"5. La mayor estatura entre los estudiantes varones: {max_alt_varones}")

# 6. La menor estatura entre las estudiantes
min_alt_mujeres = df[df['Gender'] == 'F']['Height'].min()
print(f"6. La menor estatura entre las estudiantes: {min_alt_mujeres}")

# 7. El peso más ligero entre los varones de estatura superior a 1.75m
varones_altos = df[(df['Gender'] == 'M') & (df['Height'] > 175.0)]
peso_min_varones_altos = varones_altos['Weight'].min()
print(f"7. El peso más ligero entre los estudiantes varones de estatura superior a 1.75 m: {peso_min_varones_altos}")

# 8. El mayor peso entre las estudiantes de estatura inferior a 1.60 m (160.0 cm)
mujeres_bajas = df[(df['Gender'] == 'F') & (df['Height'] < 160.0)]
peso_max_mujeres_bajas = mujeres_bajas['Weight'].max() if not mujeres_bajas.empty else "No hay estudiantes con esas características"
print(f"8. El mayor peso entre las estudiantes de estatura inferior a 1.60 m: {peso_max_mujeres_bajas}")

# 9. La nota promedio de los alumnos (hombres y mujeres) sin ausencias ('N')
nota_prom_sin_ausencias = df[df['Absence'] == 'N']['Grade'].mean()
print(f"9. La nota promedio de los alumnos sin ausencias: {nota_prom_sin_ausencias:.2f}")

# 10. La nota promedio de los alumnos (hombres y mujeres) con absentismo ('Y')
nota_prom_con_absentismo = df[df['Absence'] == 'Y']['Grade'].mean()
print(f"10. Nota promedio con absentismo ('Y'): {nota_prom_con_absentismo:.1f}")

# 11. Altura promedio de los alumnos (hombres y mujeres) con grupo sanguíneo 'A' o 'AB'
alt_prom_A_AB = df[df['Bloodtype'].isin(['A', 'AB'])]['Height'].mean()
print(f"11. Altura promedio con grupo sanguíneo 'A' o 'AB': {alt_prom_A_AB:.1f} cm")

# 12. Altura promedio de los alumnos varones con grupo sanguíneo 'A' o 'AB'
varones_A_AB = df[(df['Gender'] == 'M') & (df['Bloodtype'].isin(['A', 'AB']))]
alt_prom_varones_A_AB = varones_A_AB['Height'].mean()
print(f"12. Altura promedio de varones con grupo 'A' o 'AB': {alt_prom_varones_A_AB:.1f} cm")

# 13. Edad promedio de los alumnos (hombres y mujeres) con absentismo ('Y') cuya nota es mayor o igual a 3
edad_prom_abs_nota = df[(df['Absence'] == 'Y') & (df['Grade'] >= 3)]['Age'].mean()
print(f"13. Edad promedio con absentismo ('Y') y nota >= 3: {edad_prom_abs_nota:.1f} años")
```

```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
%matplotlib inline
# Asegúrate de encontrarte en el directorio donde se encuentra el archivo.
# os.chdir(r'~~') # Reemplaza el path con el tuyo.
df = pd.read_csv('../data/data_coffeeshop.csv', header='infer', na_values=[' '])
print(df.shape)
df.head(5)
```

1) Genera una tabla de frecuencias para la variable 'yearOfStart' y visualízala por año.
• Ordena por año.
• Dibuja una gráfica de línea de 1997 a 2014. <= Pista: plt.xlim()
2) Divide los datos por el estado actual del negocio ('In' o 'Out'). Después, visualiza la tendencia annual de las
frecuencias de la variable 'yearOfStart'.
• Ordena por año.
• Dibuja dos gráficas de líneas superpuestas de 1997 a 2014.
• Usa el objeto 'figure'.
3) A partir de los resultados de 1) y 2), ¿cómo describirías la tendencia en general?

```python
#1
# Generar la tabla de frecuencias para 'Year of Start' y ordenar por año
frecuencias_anio = df['Year of Start'].value_counts().sort_index()

# Dibujar la gráfica de línea
plt.figure(figsize=(10, 6))
plt.plot(frecuencias_anio.index, frecuencias_anio.values, marker='o', linestyle='-')

# Limitar el eje X de 1997 a 2014 como indica la pista
plt.xlim(1997, 2014)

# Formato adicional para que se vea bien
plt.title('Frecuencia de inicio de negocios por año')
plt.xlabel('Año de inicio (Year of Start)')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()
```

```python
#2
# Dividir los datos por el estado actual del negocio
df_in = df[df['Current State'] == 'In']
df_out = df[df['Current State'] == 'Out']

# Generar y ordenar las frecuencias anuales para cada estado
frecuencias_in = df_in['Year of Start'].value_counts().sort_index()
frecuencias_out = df_out['Year of Start'].value_counts().sort_index()

# Usar el objeto 'figure' y dibujar las gráficas superpuestas
fig = plt.figure(figsize=(10, 6))

plt.plot(frecuencias_in.index, frecuencias_in.values, label='In (Activos)', marker='o', color='blue')
plt.plot(frecuencias_out.index, frecuencias_out.values, label='Out (Cerrados)', marker='x', color='red')

# Limitar el eje X de 1997 a 2014
plt.xlim(1997, 2014)

plt.title('Tendencia anual de frecuencias por Estado Actual (In vs Out)')
plt.xlabel('Año de inicio (Year of Start)')
plt.ylabel('Frecuencia')
plt.legend()
plt.grid(True)
plt.show()
```

## Descripción de lo que pasa en ambas gráficas

En el gráfico 1, la tendencia más alta fue en el año 2014 y a excepción del año 2003 al año 2006, siempre aumentó.

En el gráfico 2, los activos siempre fueron en aumento con el pasar de los años y los out tuvieron un pico en 2011 y de ahí fueron en disminución.
